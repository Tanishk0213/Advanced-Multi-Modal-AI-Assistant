```python
from pathlib import Path
from typing import List
import os
import google.generativeai as genai

from langchain.document_loaders import TextLoader
from langchain.document_loaders.csv_loader import CSVLoader
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS

BASE_DIR = Path("/content")
VECTOR_DB_PATH = BASE_DIR / "faiss_index"
BASE_DATASET_PATH = BASE_DIR / "dataset.csv"
KNOWLEDGE_SOURCES_DIR = BASE_DIR / "knowledge_sources"

genai.configure(api_key=os.environ["GOOGLE_API_KEY"])
model = genai.GenerativeModel("gemini-2.5-flash")

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

def ensure_directories():
    KNOWLEDGE_SOURCES_DIR.mkdir(exist_ok=True)

def load_base_csv():
    loader = CSVLoader(
        file_path=str(BASE_DATASET_PATH),
        source_column="prompt",
        encoding="latin-1"
    )
    return loader.load()

def load_extra_documents():
    docs = []
    if not KNOWLEDGE_SOURCES_DIR.exists():
        return docs

    for file_path in KNOWLEDGE_SOURCES_DIR.iterdir():
        if file_path.suffix.lower() == ".txt":
            loader = TextLoader(str(file_path), encoding="utf-8")
            docs.extend(loader.load())
        elif file_path.suffix.lower() == ".csv":
            loader = CSVLoader(file_path=str(file_path), encoding="latin-1")
            docs.extend(loader.load())

    return docs

def get_all_documents():
    docs = []
    docs.extend(load_base_csv())
    docs.extend(load_extra_documents())
    return docs

def create_vector_db():
    ensure_directories()
    docs = get_all_documents()
    vectordb = FAISS.from_documents(docs, embedding_model)
    vectordb.save_local(str(VECTOR_DB_PATH))
    print("✅ Knowledge Base Created")

def refresh_vector_db():
    create_vector_db()
    print("✅ Knowledge Base Updated")

def ask_question(question):
    vectordb = FAISS.load_local(
        str(VECTOR_DB_PATH),
        embedding_model
    )

    docs = vectordb.similarity_search(question, k=3)
    context = "\n\n".join([doc.page_content for doc in docs])

    prompt = f"""
You are a helpful assistant.
Answer only from the given context.
If the answer is not present, say: I don't know.

Context:
{context}

Question:
{question}
"""

    response = model.generate_content(prompt)
    return response.text
```

```python
import xml.etree.ElementTree as ET
import pandas as pd

def parse_medquad_xml(xml_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()

    rows = []

    for qa in root.findall(".//QAPair"):
        question = ""
        answer = ""

        q_node = qa.find("Question")
        a_node = qa.find("Answer")

        if q_node is not None:
            question = " ".join(q_node.itertext()).strip()

        if a_node is not None:
            answer = " ".join(a_node.itertext()).strip()

        if question and answer:
            rows.append({
                "question": question,
                "answer": answer
            })

    return pd.DataFrame(rows)
```

```python
df = parse_medquad_xml("/content/0000001.xml")

print("Rows:", len(df))
df.head()
```

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

questions = df["question"].tolist()
embeddings = model.encode(questions).astype("float32")

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)

print("Index size:", index.ntotal)
```

```python
def medical_qa(query):
    q_emb = model.encode([query]).astype("float32")
    D, I = index.search(q_emb, 1)

    return df.iloc[I[0][0]]["answer"]
```

```python
print(medical_qa("What is Acromegaly?"))
```

```python
def get_research_papers(topic):

    search = arxiv.Search(
        query=topic,
        max_results=3
    )

    results = list(
        client.results(search)
    )

    papers = []

    for result in results:

        papers.append({
            "title": result.title,
            "summary": result.summary
        })

    return papers
```

```python
def research_chatbot(query):

    papers = get_research_papers(query)

    context = ""

    for p in papers:

        context += f"""
Title:
{p['title']}

Summary:
{p['summary']}
"""

    prompt = f"""
You are an expert research assistant.

Using these research papers,
answer the user's question clearly.

Research Papers:
{context}

Question:
{query}

Explain in simple words.
"""

    response = gemini_model.generate_content(prompt)

    return response.text
```

```python
def detect_sentiment(text):

    polarity = TextBlob(
        text
    ).sentiment.polarity

    if polarity > 0:
        return "positive"

    elif polarity < 0:
        return "negative"

    else:
        return "neutral"
```

```python
print(
    detect_sentiment(
        "I am very happy today"
    )
)
```

```python
print(
    detect_sentiment(
        "I am feeling very sad"
    )
)
```

```python
def emotional_chatbot(message):

    sentiment = detect_sentiment(message)

    if sentiment == "positive":

        return "I'm glad to hear that! 😊"

    elif sentiment == "negative":

        return "I'm sorry you're facing this issue. How can I help? 💙"

    else:

        return "I understand. Tell me more."
```

```python
print(
    emotional_chatbot(
        "I failed my exam"
    )
)
```

```python
def smart_sentiment_chat(message):

    sentiment = detect_sentiment(message)

    prompt = f"""
User sentiment:
{sentiment}

User message:
{message}

Respond appropriately based on emotion.
"""

    response = gemini_model.generate_content(prompt)

    return response.text
```

```python
print(
    smart_sentiment_chat(
        "I feel stressed about exams"
    )
)
```

```python
def detect_language(text):

    return detect(text)
```

```python
language_map = {
    "ta": "Tamil",
    "en": "English",
    "hi": "Hindi",
    "fr": "French",
    "de": "German",
    "te": "Telugu",
    "ml": "Malayalam"
}
def translate_text(text, target_lang):

    translated = GoogleTranslator(
        source='auto',
        target=target_lang
    ).translate(text)

    return translated
```

```python
lang = detect_language(
    "நீங்கள் எப்படி இருக்கிறீர்கள்"
)

print(
    language_map.get(lang, lang)
)
```
